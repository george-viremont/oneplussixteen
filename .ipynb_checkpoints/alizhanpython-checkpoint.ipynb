```python
import os
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.metrics import classification_report, roc_auc_score, average_precision_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder


DATA_PATH = "flights_weather_sample.csv"


# -----------------------------
# 1. Load and prepare data
# -----------------------------

def load_data(path=DATA_PATH):
    if not os.path.exists(path):
        raise FileNotFoundError(f"Could not find {path}")

    df = pd.read_csv(path)
    return df


def add_time_columns(df):
    df = df.copy()

    df["sched_dep_dt"] = pd.to_datetime(
        df["date"] + " " + df["sched_dep_local"],
        errors="coerce"
    )

    df["scheduled_arr_dt"] = pd.to_datetime(
        df["date"] + " " + df["scheduled_arr_local"],
        errors="coerce"
    )

    # If arrival clock time is earlier than departure clock time,
    # assume overnight arrival.
    overnight = df["scheduled_arr_dt"] < df["sched_dep_dt"]
    df.loc[overnight, "scheduled_arr_dt"] += pd.Timedelta(days=1)

    return df


# -----------------------------
# 2. Build cascade features
# -----------------------------

def add_cascade_features(df):
    """
    Builds inbound-aircraft delay features from tail_number.

    Important:
    We only use the PREVIOUS leg's delay information.
    We do not use the current row's dep_delay_min, arr_delay_min,
    delayed_15, delay_cause, etc. as model inputs.
    """

    df = df.copy()
    df = df.sort_values(["tail_number", "sched_dep_dt"]).reset_index(drop=True)

    g = df.groupby("tail_number", sort=False)

    df["prev_flight_id"] = g["flight_id"].shift(1)
    df["prev_arr_delay_min"] = g["arr_delay_min"].shift(1)
    df["prev_dep_delay_min"] = g["dep_delay_min"].shift(1)
    df["prev_sched_arr_dt"] = g["scheduled_arr_dt"].shift(1)

    df["turnaround_buffer_min"] = (
        (df["sched_dep_dt"] - df["prev_sched_arr_dt"])
        .dt.total_seconds() / 60.0
    )

    # The sample has local times but not full timezone-normalized aircraft routing.
    # Treat impossible or very large turnarounds as unknown.
    invalid_turnaround = (
        (df["turnaround_buffer_min"] < 0) |
        (df["turnaround_buffer_min"] > 8 * 60)
    )
    df.loc[invalid_turnaround, "turnaround_buffer_min"] = np.nan

    df["prev_arr_delay_min"] = df["prev_arr_delay_min"].fillna(0)
    df["prev_dep_delay_min"] = df["prev_dep_delay_min"].fillna(0)

    median_turnaround = df["turnaround_buffer_min"].median()
    usable_buffer = df["turnaround_buffer_min"].fillna(median_turnaround)

    df["inbound_delay_after_buffer_min"] = np.maximum(
        df["prev_arr_delay_min"] - usable_buffer,
        0
    )

    df["had_late_inbound"] = (df["prev_arr_delay_min"] >= 15).astype(int)
    df["tight_turnaround"] = (df["turnaround_buffer_min"] <= 45).astype(int)

    return df


# -----------------------------
# 3. Build congestion features
# -----------------------------

def add_congestion_features(df):
    df = df.copy()

    df["departures_origin_hour"] = (
        df.groupby(["date", "origin", "dep_hour"])["flight_id"]
        .transform("count")
    )

    df["origin_hour_congestion_pctile"] = (
        df.groupby("origin")["departures_origin_hour"]
        .rank(pct=True)
    )

    df["is_early_bank"] = df["dep_hour"].between(5, 8).astype(int)
    df["is_evening_bank"] = df["dep_hour"].between(16, 20).astype(int)

    return df


# -----------------------------
# 4. Build weather features
# -----------------------------

def add_weather_features(df):
    df = df.copy()

    df["is_snow_code"] = df["weather_code"].between(71, 77).astype(int)
    df["is_rain_code"] = df["weather_code"].between(51, 67).astype(int)
    df["is_shower_code"] = df["weather_code"].between(80, 82).astype(int)
    df["is_fog_code"] = (df["weather_code"] == 45).astype(int)
    df["is_thunderstorm_code"] = (df["weather_code"] == 95).astype(int)

    df["weather_severity_score"] = (
        2.5 * df["snowfall_cm"].fillna(0)
        + 1.5 * df["precip_mm"].fillna(0)
        + 0.03 * df["wind_gust_kmh"].fillna(0)
        + 2.0 * df["is_fog_code"]
        + 3.0 * df["is_thunderstorm_code"]
    )

    return df


def build_feature_table(df):
    df = add_time_columns(df)
    df = add_cascade_features(df)
    df = add_congestion_features(df)
    df = add_weather_features(df)

    return df.sort_values("sched_dep_dt").reset_index(drop=True)


# -----------------------------
# 5. Train scikit-learn model
# -----------------------------

def build_model():
    numeric_features = [
        "day_of_week",
        "dep_hour",
        "distance_km",

        "temp_c",
        "wind_speed_kmh",
        "wind_gust_kmh",
        "precip_mm",
        "snowfall_cm",
        "cloud_cover_pct",
        "weather_code",

        "prev_arr_delay_min",
        "prev_dep_delay_min",
        "turnaround_buffer_min",
        "inbound_delay_after_buffer_min",
        "had_late_inbound",
        "tight_turnaround",

        "departures_origin_hour",
        "origin_hour_congestion_pctile",
        "is_early_bank",
        "is_evening_bank",

        "is_snow_code",
        "is_rain_code",
        "is_shower_code",
        "is_fog_code",
        "is_thunderstorm_code",
        "weather_severity_score",
    ]

    categorical_features = [
        "carrier",
        "origin",
        "dest",
    ]

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "num",
                SimpleImputer(strategy="median"),
                numeric_features
            ),
            (
                "cat",
                Pipeline(
                    steps=[
                        ("imputer", SimpleImputer(strategy="most_frequent")),
                        ("onehot", OneHotEncoder(handle_unknown="ignore")),
                    ]
                ),
                categorical_features
            ),
        ]
    )

    classifier = RandomForestClassifier(
        n_estimators=500,
        max_depth=8,
        min_samples_leaf=15,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1,
    )

    model = Pipeline(
        steps=[
            ("preprocess", preprocessor),
            ("model", classifier),
        ]
    )

    feature_columns = numeric_features + categorical_features

    return model, feature_columns


# -----------------------------
# 6. Explanation layer
# -----------------------------

def explain_flight(row, risk):
    reasons = []
    actions = []

    if row["prev_arr_delay_min"] >= 15:
        reasons.append(
            f"Inbound aircraft arrived {row['prev_arr_delay_min']:.0f} min late"
        )
        actions.append(
            "Check whether an aircraft swap is possible"
        )

    if pd.notna(row["turnaround_buffer_min"]) and row["turnaround_buffer_min"] <= 45:
        reasons.append(
            f"Tight turnaround: only {row['turnaround_buffer_min']:.0f} min scheduled buffer"
        )
        actions.append(
            "Prioritize gate, cleaning, fueling, and boarding resources"
        )

    if row["inbound_delay_after_buffer_min"] >= 10:
        reasons.append(
            f"Estimated unrecovered inbound delay: {row['inbound_delay_after_buffer_min']:.0f} min"
        )
        actions.append(
            "Escalate to hub controller for recovery decision"
        )

    if row["snowfall_cm"] > 0:
        reasons.append(
            f"Snowfall at {row['origin']}: {row['snowfall_cm']:.1f} cm during departure hour"
        )
        actions.append(
            "Prepare de-icing and winter ramp plan"
        )

    if row["precip_mm"] >= 1:
        reasons.append(
            f"Precipitation at {row['origin']}: {row['precip_mm']:.1f} mm during departure hour"
        )
        actions.append(
            "Expect slower ramp and taxi operations"
        )

    if row["wind_gust_kmh"] >= 45:
        reasons.append(
            f"High wind gusts at {row['origin']}: {row['wind_gust_kmh']:.0f} km/h"
        )
        actions.append(
            "Monitor ground-handling and runway-flow restrictions"
        )

    if row["is_fog_code"] == 1:
        reasons.append("Fog / visibility risk at origin")
        actions.append("Monitor ATC flow restrictions")

    if row["is_thunderstorm_code"] == 1:
        reasons.append("Thunderstorm risk at origin")
        actions.append("Prepare reroute or delay-program decision")

    if row["origin_hour_congestion_pctile"] >= 0.85:
        reasons.append(
            f"Busy {row['origin']} departure bank: "
            f"{row['departures_origin_hour']:.0f} flights scheduled this hour"
        )
        actions.append(
            "Review pushback sequence and ATC slot risk"
        )

    if row["is_evening_bank"] == 1:
        reasons.append(
            "Evening departure bank: network delays are more likely to have accumulated"
        )

    if len(reasons) == 0:
        reasons.append(
            "No single dominant operational driver found; risk is mainly from learned carrier, route, and time patterns"
        )
        actions.append(
            "Monitor normally; no immediate intervention suggested"
        )

    actions = list(dict.fromkeys(actions))

    if risk >= 0.70:
        priority = "HIGH"
    elif risk >= 0.40:
        priority = "MEDIUM"
    else:
        priority = "LOW"

    return {
        "flight_id": row["flight_id"],
        "route": f"{row['origin']}->{row['dest']}",
        "scheduled_departure": row["sched_dep_local"],
        "delay_risk": round(float(risk), 3),
        "priority": priority,
        "reasons": reasons[:5],
        "recommended_actions": actions[:4],
    }


# -----------------------------
# 7. Run full algorithm
# -----------------------------

def main():
    df_raw = load_data(DATA_PATH)
    df = build_feature_table(df_raw)

    # Drop cancellations and rows where delayed_15 is missing.
    # You can build a separate cancellation model later.
    modeling_df = df[
        (df["cancelled"] == 0) &
        (df["delayed_15"].notna())
    ].copy()

    modeling_df["delayed_15"] = modeling_df["delayed_15"].astype(int)

    model, feature_columns = build_model()

    # Time-based split: train on earlier flights, test on later flights.
    # This is better than random split for operational prediction.
    split_idx = int(len(modeling_df) * 0.80)

    train_df = modeling_df.iloc[:split_idx]
    test_df = modeling_df.iloc[split_idx:]

    X_train = train_df[feature_columns]
    y_train = train_df["delayed_15"]

    X_test = test_df[feature_columns]
    y_test = test_df["delayed_15"]

    model.fit(X_train, y_train)

    risk_scores = model.predict_proba(X_test)[:, 1]
    predictions = (risk_scores >= 0.50).astype(int)

    print("Rows used:", len(modeling_df))
    print("Delay base rate:", round(modeling_df["delayed_15"].mean(), 3))
    print("ROC-AUC:", round(roc_auc_score(y_test, risk_scores), 3))
    print("Average precision:", round(average_precision_score(y_test, risk_scores), 3))
    print()
    print(classification_report(y_test, predictions, digits=3))

    # Score test flights
    scored = test_df.copy()
    scored["delay_risk"] = risk_scores

    # Show top-risk flights
    top_risk = scored.sort_values("delay_risk", ascending=False).head(10)

    print("\nTop delay-risk flights:\n")

    for _, row in top_risk.iterrows():
        alert = explain_flight(row, row["delay_risk"])

        print(
            f"{alert['priority']} | {alert['flight_id']} | "
            f"{alert['route']} | {alert['scheduled_departure']} | "
            f"risk={alert['delay_risk']:.1%}"
        )

        print("Reasons:")
        for reason in alert["reasons"]:
            print(f" - {reason}")

        print("Recommended actions:")
        for action in alert["recommended_actions"]:
            print(f" - {action}")

        print()

    # Save scored output
    output_cols = [
        "flight_id",
        "date",
        "carrier",
        "origin",
        "dest",
        "sched_dep_local",
        "delay_risk",
        "delayed_15",
        "delay_cause",

        "prev_arr_delay_min",
        "turnaround_buffer_min",
        "inbound_delay_after_buffer_min",

        "snowfall_cm",
        "precip_mm",
        "wind_gust_kmh",

        "departures_origin_hour",
        "origin_hour_congestion_pctile",
    ]

    scored[output_cols].sort_values(
        "delay_risk",
        ascending=False
    ).to_csv(
        "delay_risk_scored_flights.csv",
        index=False
    )

    print("Saved scored file: delay_risk_scored_flights.csv")


if __name__ == "__main__":
    main()
```
